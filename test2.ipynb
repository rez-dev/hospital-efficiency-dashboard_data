```python
import pandas as pd
import statsmodels.api as sm
from statsmodels.miscmodels.ordinal_model import OrderedModel

# 1. Simular dataset
data = pd.DataFrame({
    'Eficiencia': [0.83, 0.65, 0.91, 0.49, 0.75],
    'Region': ['Metropolitana', 'Valparaíso', 'Biobío', 'Metropolitana', 'Valparaíso'],
    'Complejidad': ['Alta', 'Media', 'Alta', 'Baja', 'Media']
})

# 2. Codificar variables
data['Complejidad'] = data['Complejidad'].map({'Baja': 1, 'Media': 2, 'Alta': 3})  # ordinal
data = pd.get_dummies(data, columns=['Region'], drop_first=True)  # dummies para Región

# 3. Variables
X = data[['Complejidad', 'Region_Metropolitana', 'Region_Valparaíso']].copy()
# Convert boolean columns to int
X['Region_Metropolitana'] = X['Region_Metropolitana'].astype(int)
X['Region_Valparaíso'] = X['Region_Valparaíso'].astype(int)
y = data['Eficiencia']

# 4. Modelo Tobit aproximado usando regresión lineal truncada (simplificado)
X = sm.add_constant(X)
model = sm.OLS(y, X).fit()

print(model.summary())
```
